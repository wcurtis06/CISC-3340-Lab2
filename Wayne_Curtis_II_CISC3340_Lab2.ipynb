## Worked Example: Rectangle Perimeter

This example is filled in completely so you can see the expected format
before you start on your own. It asks the user for the length and width
of a rectangle and outputs the perimeter.


**Pseudocode**

```
START
  INPUT length
  INPUT width
  SET perimeter = 2 * (length + width)
  OUTPUT perimeter
END
```

### Question 6: Minutes to Hours and Minutes

Write a program that asks the user for a number of minutes and outputs how many full hours and how many leftover minutes that equals. For example, 130 minutes should output 2 hours and 10 minutes.

- Output both values, either together or in two separate OUTPUT steps.

```
START
  INPUT total_minutes
```

### Question 9: Pass or Fail

Write a program that asks the user for a test score and outputs Pass if the score is 60 or higher, and Fail otherwise. This is the same example we broke down in class.

**Requirements**
- Use exactly one IF, one ELSE, and one ENDIF in your pseudocode.

Write a program that asks the user to enter a password and outputs Strong enough if the password is at least 8 characters long, and Too short otherwise.
